```python
# code to draw histogram distribution of the current salinity

# import packages
import openvisuspy as ovp

import matplotlib
matplotlib.use("Agg")

import matplotlib.pyplot as plt
import numpy as np
```

```python
# import and load dataset
salinity ="pelican://osg-htc.org/nasa/nsdf/climate1/llc4320/idx/salt/salt_llc4320_x_y_depth.idx"

print(salinity)

s_ds = ovp.LoadDataset(salinity)
```

```python
# import data range and time index
# north atlantic
x_range=[14000, 17200]
y_range=[8000, 11000]
z_range=[0,90]

time = 2000

quality = -9
```

```python
# read the dataset
S = s_ds.db.read(time=time, x=x_range, y=y_range, z=z_range, quality = quality)

print("S shape:", S.shape)

S = S.astype(np.float32)
```

```python
# apply ocean mask
ocean_mask = (S != 0)
S = np.where(ocean_mask, S, np.nan)
```

```python
# plotting histogram

# remove land / invalid values
S_valid = S[np.isfinite(S) & (S != 0)]

# define Percentiles
p25 = np.percentile(S_valid, 8)
p50 = np.percentile(S_valid, 40)
p75 = np.percentile(S_valid, 92)

print(
    f"t={time}: "
    f"25th percentile = {p25: 4f},"
    f"50th percentile = {p50:.4f}, "
    f"75th percentile = {p75:.4f}"
)

# Create separate histogram
fig, ax = plt.subplots(figsize=(9, 6))

ax.hist(
    S_valid,
    bins=np.linspace(30, 40, 101),
    density=True,
    edgecolor="black",
    alpha=0.8,
)

ax.axvline(
    p25,
    color="red",
    linestyle="-.",
    linewidth=2,
    label=f"25th percentile = {p25:.3f}"
    )

ax.axvline(
    p50,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"50th percentile = {p50:.3f}"
)

ax.axvline(
    p75,
    color="red",
    linestyle=":",
    linewidth=2,
    label=f"75th percentile = {p75:.3f}"
)

ax.set_xlabel("Salinity (g kg$^{-1}$)")
ax.set_ylabel("Probability Density")
ax.set_title(f"Salinity Distribution — Time Step {time}")

ax.set_xlim(30, 40)

ax.legend()
ax.grid(alpha=0.3)

fig.tight_layout()

fig.savefig(
    f"salinity_histogram_t{time}.png",
    dpi=150,
    bbox_inches="tight"
)

print(f"Saved salinity_histogram_t{time}.png", flush=True)
```
